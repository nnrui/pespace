```python
import numpy as np
from matplotlib import pyplot as plt
%matplotlib inline
import taichi as ti
ti.init(arch=ti.cuda, default_fp=ti.f64, cpu_max_num_threads=1, offline_cache=False, debug=True)

from pespace.detector.noise import available_noise_models
```

```python
f_min = 1e-5
f_max = 1
nsample = 2**20
dt = 0.5
freqs = np.fft.rfftfreq(nsample, dt)
freqs = freqs[(freqs>=f_min) * (freqs<=f_max)]

channels = ("X", "Y", "Z", "A", "E", "T")
```

```python
LISA_SciRDv1 = available_noise_models['LISA_SciRDv1']
Taiji_TDC = available_noise_models['Taiji_TDC']
Tianqin_GWSpace = available_noise_models['Tianqin_GWSpace']

psd_lisa_tdi1 = LISA_SciRDv1(freqs, channels, False, "1.5")
psd_lisa_tdi2 = LISA_SciRDv1(freqs, channels, False, "2.0")

psd_taiji_tdi1 = Taiji_TDC(freqs, channels, False, "1.5")
psd_taiji_tdi2 = Taiji_TDC(freqs, channels, False, "2.0")

psd_tianqin_tdi1 = Tianqin_GWSpace(freqs, channels, False, "1.5")
psd_tianqin_tdi2 = Tianqin_GWSpace(freqs, channels, False, "2.0")
```

```python
plt.figure()
plt.title("X (Y, Z) channel, TDI 1.5")
plt.loglog(freqs, psd_lisa_tdi1['X'], label='LISA_SciRDv1')
plt.loglog(freqs, psd_taiji_tdi1['X'], label='Taiji_TDC')
plt.loglog(freqs, psd_tianqin_tdi1['X'], label='Tianqin_GWSpace')
plt.grid(True)
plt.xlim(f_min, f_max)
plt.legend()
plt.show()

plt.figure()
plt.title("X (Y, Z) channel, TDI 2.0")
plt.loglog(freqs, psd_lisa_tdi2['X'], label='LISA_SciRDv1')
plt.loglog(freqs, psd_taiji_tdi2['X'], label='Taiji_TDC')
plt.loglog(freqs, psd_tianqin_tdi2['X'], label='Tianqin_GWSpace')
plt.grid(True)
plt.xlim(f_min, f_max)
plt.legend()
plt.show()


```

```python
plt.figure()
plt.title("A (E) channel, TDI 1.5")
plt.loglog(freqs, psd_lisa_tdi1['A'], label='LISA_SciRDv1')
plt.loglog(freqs, psd_taiji_tdi1['A'], label='Taiji_TDC')
plt.loglog(freqs, psd_tianqin_tdi1['A'], label='Tianqin_GWSpace')
plt.grid(True)
plt.xlim(f_min, f_max)
plt.legend()
plt.show()

plt.figure()
plt.title("T channel, TDI 1.5")
plt.loglog(freqs, psd_lisa_tdi1['T'], label='LISA_SciRDv1')
plt.loglog(freqs, psd_taiji_tdi1['T'], label='Taiji_TDC')
plt.loglog(freqs, psd_tianqin_tdi1['T'], label='Tianqin_GWSpace')
plt.grid(True)
plt.xlim(f_min, f_max)
plt.legend()
plt.show()

plt.figure()
plt.title("A (E) channel, TDI 2.0")
plt.loglog(freqs, psd_lisa_tdi2['A'], label='LISA_SciRDv1')
plt.loglog(freqs, psd_taiji_tdi2['A'], label='Taiji_TDC')
plt.loglog(freqs, psd_tianqin_tdi2['A'], label='Tianqin_GWSpace')
plt.grid(True)
plt.xlim(f_min, f_max)
plt.legend()
plt.show()

plt.figure()
plt.title("T channel, TDI 2.0")
plt.loglog(freqs, psd_lisa_tdi2['T'], label='LISA_SciRDv1')
plt.loglog(freqs, psd_taiji_tdi2['T'], label='Taiji_TDC')
plt.loglog(freqs, psd_tianqin_tdi2['T'], label='Tianqin_GWSpace')
plt.grid(True)
plt.xlim(f_min, f_max)
plt.legend()
plt.show()
```
